FININT – Phase 2: Data Preparation

```python
# Create a modelling copy
model_data = credit_card.copy()

# Remove descriptive columns that are not original model features
model_data = model_data.drop(
    columns=["Transaction_Type", "Transaction_Hour"],
    errors="ignore"
)

# Separate input features and target
X = model_data.drop(columns=["Class"])
y = model_data["Class"]

print("Feature dataset:", X.shape)
print("Target dataset:", y.shape)
```

```python
duplicate_mask = model_data.duplicated(
    subset=X.columns,
    keep=False
)

duplicate_label_check = (
    model_data.loc[duplicate_mask]
    .groupby(list(X.columns))["Class"]
    .nunique()
)

conflicting_groups = duplicate_label_check[
    duplicate_label_check > 1
]

print("Duplicate rows involved:", duplicate_mask.sum())
print("Conflicting duplicate groups:", len(conflicting_groups))
```

```python
if len(conflicting_groups) == 0:
    model_data = model_data.drop_duplicates(
        subset=X.columns
    ).reset_index(drop=True)

    print("Dataset after duplicate removal:", model_data.shape)
else:
    print("Conflicting labels found. Investigate before removal.")
```

```python
X = model_data.drop(columns=["Class"])
y = model_data["Class"]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("\nTraining class distribution:")
print(y_train.value_counts())

print("\nTesting class distribution:")
print(y_test.value_counts())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train = X_train.copy()
X_test = X_test.copy()

X_train[["Time", "Amount"]] = scaler.fit_transform(
    X_train[["Time", "Amount"]]
)

X_test[["Time", "Amount"]] = scaler.transform(
    X_test[["Time", "Amount"]]
)
```

```python
from sklearn.linear_model import LogisticRegression

logistic_model = LogisticRegression(
    class_weight="balanced",
    max_iter=1000,
    random_state=42
)

logistic_model.fit(X_train, y_train)

print("Logistic Regression training completed.")
```

```python
from sklearn.ensemble import RandomForestClassifier

validation_model = RandomForestClassifier(
    n_estimators=300,
    class_weight="balanced_subsample",
    max_features="sqrt",
    random_state=42,
    n_jobs=-1
)

validation_model.fit(X_fit, y_fit)

val_probabilities = validation_model.predict_proba(X_val)[:, 1]

print("Validation predictions generated.")
```

```python
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score
)

thresholds = [0.10, 0.20, 0.30, 0.40, 0.50, 0.60, 0.70, 0.80, 0.90]

threshold_results = []

for threshold in thresholds:

    predictions = (
        val_probabilities >= threshold
    ).astype(int)

    threshold_results.append({
        "Threshold": threshold,
        "Precision": precision_score(
            y_val, predictions, zero_division=0
        ),
        "Recall": recall_score(
            y_val, predictions, zero_division=0
        ),
        "F1_Score": f1_score(
            y_val, predictions, zero_division=0
        ),
        "Fraud_Alerts": predictions.sum()
    })

threshold_analysis = pd.DataFrame(threshold_results)

threshold_analysis.round(3)
```

```python
def assess_transaction(transaction, model, explainer, threshold=0.20):

    probability = model.predict_proba(transaction)[0][1]

    prediction = (
        "Fraudulent"
        if probability >= threshold
        else "Legitimate"
    )

    explanation = explainer(transaction)

    shap_values = explanation[0, :, 1].values

    contributions = pd.DataFrame({
        "Feature": transaction.columns,
        "Feature_Value": transaction.iloc[0].values,
        "SHAP_Contribution": shap_values
    })

    contributions = contributions.sort_values(
        by="SHAP_Contribution",
        key=lambda values: values.abs(),
        ascending=False
    )

    return {
        "Fraud_Probability": round(float(probability), 4),
        "Predicted_Class": prediction,
        "Threshold": threshold,
        "Top_Contributors": contributions.head(10)
    }
```

```python
transaction_report = assess_transaction(
    single_transaction,
    random_forest_model,
    explainer
)

print("Fraud probability:",
      transaction_report["Fraud_Probability"])

print("Predicted classification:",
      transaction_report["Predicted_Class"])

transaction_report["Top_Contributors"]
```

```python
# Prepare recipient activity analysis
activity_results = []

for recipient, group in recipient_history.groupby("nameDest"):

    # Sort transactions chronologically
    group = group.sort_values("step").reset_index(drop=True)

    # Locate fraudulent transaction positions
    fraud_positions = np.flatnonzero(
        group["isFraud"].to_numpy() == 1
    )

    first_fraud_position = fraud_positions[0]
    second_fraud_position = fraud_positions[1]

    # Separate legitimate transactions into three periods
    before = group.iloc[:first_fraud_position]

    between = group.iloc[
        first_fraud_position + 1:second_fraud_position
    ]

    after = group.iloc[second_fraud_position + 1:]

    # Collect the results
    activity_results.append({
        "Recipient_Account": recipient,

        "Legitimate_Before_Fraud": len(before),
        "Amount_Before_Fraud": before.loc[
            before["isFraud"] == 0, "amount"
        ].sum(),

        "Legitimate_Between_Fraud": len(between),
        "Amount_Between_Fraud": between.loc[
            between["isFraud"] == 0, "amount"
        ].sum(),

        "Legitimate_After_Fraud": len(after),
        "Amount_After_Fraud": after.loc[
            after["isFraud"] == 0, "amount"
        ].sum()
    })

# Convert results into a DataFrame
recipient_sequence_analysis = pd.DataFrame(activity_results)

display(
    recipient_sequence_analysis.style.format({
        "Amount_Before_Fraud": "{:,.2f}",
        "Amount_Between_Fraud": "{:,.2f}",
        "Amount_After_Fraud": "{:,.2f}"
    })
)
```

```python
# Prepare recipient-hour transaction records
recipient_velocity = recipient_hourly[
    ["nameDest", "step", "Transactions_In_Hour"]
].copy()

# Assign a numerical code to each recipient
recipient_codes, unique_recipients = pd.factorize(
    recipient_velocity["nameDest"],
    sort=True
)

recipient_velocity["_recipient_code"] = recipient_codes

# Create a unique numerical key for each recipient-hour combination
step_base = int(recipient_velocity["step"].max()) + 1

recipient_velocity["_key"] = (
    recipient_velocity["_recipient_code"].astype("int64") * step_base
    + recipient_velocity["step"].astype("int64")
)

# Sort chronologically within each recipient
recipient_velocity = recipient_velocity.sort_values("_key").reset_index(drop=True)

# Extract numerical arrays
keys = recipient_velocity["_key"].to_numpy()
steps = recipient_velocity["step"].to_numpy()
codes = recipient_velocity["_recipient_code"].to_numpy()
hourly_counts = recipient_velocity["Transactions_In_Hour"].to_numpy()

# Build cumulative transaction counts
cumulative_counts = np.concatenate([
    [0],
    np.cumsum(hourly_counts, dtype=np.int64)
])

# Calculate historical transaction velocity
for hours in [1, 3, 6, 24]:

    # Earliest step included in the historical window
    lower_keys = (
        codes * step_base
        + np.maximum(steps - hours, 1)
    )

    # Locate the beginning of each historical window
    start_positions = np.searchsorted(
        keys,
        lower_keys,
        side="left"
    )

    # Calculate transactions received before the current hour
    recipient_velocity[f"Transactions_Previous_{hours}h"] = (
        cumulative_counts[np.arange(len(keys))]
        - cumulative_counts[start_positions]
    )

# Display results
display(recipient_velocity.head(15))

print(
    "\nTotal recipient-hour records:",
    len(recipient_velocity)
)
```
